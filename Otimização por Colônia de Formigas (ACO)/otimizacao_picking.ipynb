# Otimização da separação de pedidos em Centro de Distribuição

**Disciplina:** IA na Gestão de Negócios — PUCPR
**Empresa objeto:** Rede Vetor Varejo S.A. (caso fictício representativo)

Este notebook implementa e testa o pipeline proposto para o **ciclo operacional**
da separação de pedidos (*order picking*):

| Etapa | Técnica | Subproblema |
|---|---|---|
| 1 | Grafo do CD + Floyd–Warshall | matriz de distâncias reais |
| 2 | **Algoritmo Genético** | *Order Batching Problem* (OBP) |
| 3 | **Otimização por Colônia de Formigas (ACO)** | *Picker Routing Problem* (PRP) |

A **linha de base** é o que o WMS da empresa faz hoje: agrupamento por ordem de
chegada (FIFO) e roteirização pela heurística *S-shape*.

Três testes computacionais são executados:

1. **Didático** — um único lote, comparando as rotas S-shape e ACO.
2. **Fatorial** — uma onda completa nas quatro combinações possíveis, o que
   isola a contribuição de cada componente do pipeline.
3. **Escalabilidade** — ondas de tamanhos diferentes, com réplicas, para medir
   a estabilidade do ganho.

## 1. Configuração e parâmetros

Todos os parâmetros ficam reunidos aqui para facilitar a replicação. A semente
fixa garante que qualquer execução reproduza exatamente os números do relatório.

```python
import json
import time
from collections import defaultdict
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D

%matplotlib inline

BASE = Path.cwd()
SEMENTE = 42
rng = np.random.default_rng(SEMENTE)

# ---- Layout físico do CD ------------------------------------------------
N_RUAS = 12          # corredores de picking
N_POS = 40           # endereços por corredor
PASSO_Y = 1.20       # distância entre endereços consecutivos, em metros
PASSO_X = 3.50       # distância entre corredores vizinhos, em metros

# ---- Operação -----------------------------------------------------------
CAP_LINHAS = 18      # capacidade do carrinho de separação, em linhas
LINHAS_MIN, LINHAS_MAX = 1, 6   # linhas por pedido (perfil e-commerce)

# ---- Algoritmo Genético (batching) --------------------------------------
AG_POP = 40
AG_GERACOES = 60
AG_TORNEIO = 3
AG_P_MUT = 0.15
AG_ELITE = 2

# ---- Colônia de Formigas (roteirização) ---------------------------------
ACO_FORMIGAS = 20
ACO_ITER = 60
ACO_ALFA = 1.0       # peso do feromônio (memória coletiva)
ACO_BETA = 3.0       # peso da heurística 1/distância (visibilidade)
ACO_RHO = 0.10       # taxa de evaporação
ACO_Q = 1.0          # constante de depósito

# ---- Parâmetros de negócio (Atividade 1) --------------------------------
LINHAS_DIA = 46_000          # linhas de separação por dia no CD
CUSTO_LINHA = 0.84           # R$ por linha separada, hoje
DIAS_UTEIS = 300             # dias de operação por ano
VELOCIDADE = 1.10            # m/s de deslocamento médio do separador
FRACAO_DESLOC = 0.55         # fração do tempo de tarefa gasta em deslocamento

print(f"Endereços de picking: {N_RUAS * N_POS}")
print(f"Semente: {SEMENTE}")
```

## 2. Instância do Centro de Distribuição

O armazém é modelado como um **grafo não direcionado** $G(V, E)$:

* $V$ = doca de expedição + entradas dos corredores + endereços de *picking*;
* $E$ = trechos de corredor, ponderados pela distância real de deslocamento.

O separador só circula pelos corredores e pelas duas transversais (frontal e
traseira) — não atravessa as estruturas porta-paletes. Essa restrição é o que
torna a distância real diferente da distância euclidiana.

```python
# --- Numeração dos nós ---------------------------------------------------
# 0                      -> doca
# 1 .. N_RUAS            -> entradas frontais dos corredores
# N_RUAS+1 .. 2*N_RUAS   -> entradas traseiras
# resto                  -> endereços de picking
DOCA = 0
Y_FUNDO = (N_POS + 1) * PASSO_Y

def id_frente(r):
    return 1 + r

def id_fundo(r):
    return 1 + N_RUAS + r

def id_local(r, p):
    return 1 + 2 * N_RUAS + r * N_POS + p

N_NOS = 1 + 2 * N_RUAS + N_RUAS * N_POS

# Coordenadas (x, y) de cada nó, em metros
XY = np.zeros((N_NOS, 2))
XY[DOCA] = (-PASSO_X, 0.0)
for r in range(N_RUAS):
    XY[id_frente(r)] = (r * PASSO_X, 0.0)
    XY[id_fundo(r)] = (r * PASSO_X, Y_FUNDO)
    for p in range(N_POS):
        XY[id_local(r, p)] = (r * PASSO_X, (p + 1) * PASSO_Y)

# Vetores auxiliares: a que rua e a que posição pertence cada endereço
LOCAIS = np.array([id_local(r, p) for r in range(N_RUAS) for p in range(N_POS)])
RUA_DE = np.full(N_NOS, -1, dtype=int)
POS_DE = np.full(N_NOS, -1, dtype=int)
for r in range(N_RUAS):
    for p in range(N_POS):
        RUA_DE[id_local(r, p)] = r
        POS_DE[id_local(r, p)] = p

print(f"Nós no grafo: {N_NOS}  (1 doca + {2*N_RUAS} entradas + {len(LOCAIS)} endereços)")
print(f"Dimensões úteis: {(N_RUAS-1)*PASSO_X:.1f} m de largura x {Y_FUNDO:.1f} m de profundidade")
```

```python
# --- Matriz de adjacência ------------------------------------------------
INF = np.inf
A = np.full((N_NOS, N_NOS), INF)
np.fill_diagonal(A, 0.0)

def liga(u, v, w):
    A[u, v] = min(A[u, v], w)
    A[v, u] = A[u, v]

# Transversal frontal: doca -> rua 0 -> rua 1 -> ...
liga(DOCA, id_frente(0), PASSO_X)
for r in range(N_RUAS - 1):
    liga(id_frente(r), id_frente(r + 1), PASSO_X)
    liga(id_fundo(r), id_fundo(r + 1), PASSO_X)   # transversal traseira

# Dentro de cada corredor: entrada frontal -> endereços -> entrada traseira
for r in range(N_RUAS):
    liga(id_frente(r), id_local(r, 0), PASSO_Y)
    for p in range(N_POS - 1):
        liga(id_local(r, p), id_local(r, p + 1), PASSO_Y)
    liga(id_local(r, N_POS - 1), id_fundo(r), PASSO_Y)

print(f"Arestas não infinitas: {int(np.isfinite(A).sum() - N_NOS) // 2}")
```

```python
# --- Floyd-Warshall vetorizado ------------------------------------------
# Complexidade O(n^3), executado UMA vez, fora de linha. O laço percorre o nó
# intermediário k; a comparação de todos os pares é feita de uma vez por numpy.
t0 = time.perf_counter()
D = A.copy()
for k in range(N_NOS):
    D = np.minimum(D, D[:, k, None] + D[None, k, :])
tempo_fw = time.perf_counter() - t0

assert np.isfinite(D).all(), "grafo desconexo"
print(f"Floyd-Warshall: {N_NOS} nós em {tempo_fw:.2f} s")
print(f"Distância doca -> endereço mais distante: {D[DOCA, LOCAIS].max():.1f} m")
print(f"Distância média entre dois endereços quaisquer: "
      f"{D[np.ix_(LOCAIS, LOCAIS)].mean():.1f} m")
```

```python
def custo(seq):
    '''Distância total de uma sequência de nós, usando os caminhos mínimos.'''
    s = np.asarray(seq)
    return float(D[s[:-1], s[1:]].sum())


def caminho_xy(u, v):
    '''Polilinha realmente percorrida entre dois nós, para desenhar as rotas.

    O separador não atravessa as estruturas: ou sai pela transversal frontal,
    ou pela traseira. Escolhe-se a que resultar no menor percurso.'''
    x1, y1 = XY[u]
    x2, y2 = XY[v]
    if np.isclose(x1, x2):
        return [(x1, y1), (x2, y2)]
    por_frente = y1 + abs(x2 - x1) + y2
    por_fundo = (Y_FUNDO - y1) + abs(x2 - x1) + (Y_FUNDO - y2)
    y_via = 0.0 if por_frente <= por_fundo else Y_FUNDO
    return [(x1, y1), (x1, y_via), (x2, y_via), (x2, y2)]


# Confere que a geometria do desenho bate com a matriz de distâncias
for u, v in [(DOCA, id_local(5, 20)), (id_local(1, 3), id_local(9, 38))]:
    pts = np.array(caminho_xy(u, v))
    geo = np.abs(np.diff(pts, axis=0)).sum()
    assert np.isclose(geo, D[u, v]), (geo, D[u, v])
print("Geometria das rotas consistente com a matriz de distâncias.")
```

## 3. Geração dos pedidos

Não há acesso a dados reais de WMS, então a demanda é **gerada sinteticamente**,
respeitando a estrutura estatística observada em operações reais de varejo:

* a popularidade dos SKUs segue uma **curva ABC** (Pareto), em que uma minoria
  dos itens concentra a maior parte das linhas;
* os endereços populares estão **espalhados pelo armazém**, e não concentrados
  perto da doca — o que reproduz a linha de base descrita na Atividade 1
  (endereçamento estático por categoria mercadológica, revisto uma vez por ano);
* cada pedido tem de 1 a 6 linhas, perfil típico de comércio eletrônico.

```python
# Peso de popularidade de cada endereço: lei de potência, ranques embaralhados
ranques = rng.permutation(len(LOCAIS))
PESOS = 1.0 / np.power(ranques + 1, 1.15)
PESOS /= PESOS.sum()

acum = np.sort(PESOS)[::-1].cumsum()
n_20 = int(0.2 * len(PESOS))
print(f"Concentração: 20% dos endereços respondem por {acum[n_20]*100:.1f}% das linhas")


def gerar_pedidos(n_pedidos, gerador):
    '''Lista de pedidos; cada pedido é um array de endereços a visitar.'''
    pedidos = []
    for _ in range(n_pedidos):
        k = gerador.integers(LINHAS_MIN, LINHAS_MAX + 1)
        escolhidos = gerador.choice(LOCAIS, size=k, replace=False, p=PESOS)
        pedidos.append(np.unique(escolhidos))
    return pedidos


demo = gerar_pedidos(5, np.random.default_rng(SEMENTE))
pd.DataFrame({
    "pedido": [f"P{i+1}" for i in range(len(demo))],
    "linhas": [len(p) for p in demo],
    "ruas visitadas": [sorted(set(RUA_DE[p].tolist())) for p in demo],
})
```

## 4. Linha de base: heurística *S-shape*

É a regra embutida no WMS: o separador percorre em serpentina os corredores que
contêm itens do lote — sobe um, desce o seguinte. Simples e rápida, mas cega ao
fato de que às vezes compensa entrar e sair pelo mesmo lado.

A heurística define apenas a **ordem de visita**; a distância entre duas paradas
consecutivas é sempre o caminho mínimo real, dado pela matriz `D`.

```python
def rota_s_shape(locais):
    '''Ordem de visita em serpentina, partindo e retornando à doca.'''
    if len(locais) == 0:
        return [DOCA, DOCA]
    por_rua = defaultdict(list)
    for L in locais:
        por_rua[int(RUA_DE[L])].append(int(L))
    ruas = sorted(por_rua)

    seq = [DOCA]
    subindo = True
    for i, r in enumerate(ruas):
        itens = sorted(por_rua[r], key=lambda L: POS_DE[L], reverse=not subindo)
        if i == len(ruas) - 1:
            # No último corredor testam-se os dois sentidos e fica o melhor:
            # sem isso a linha de base ficaria artificialmente ruim.
            alt = list(reversed(itens))
            if custo(seq + alt + [DOCA]) < custo(seq + itens + [DOCA]):
                itens = alt
        seq.extend(itens)
        subindo = not subindo
    seq.append(DOCA)
    return seq


lote_demo = np.unique(np.concatenate(demo))
seq_demo = rota_s_shape(lote_demo)
print(f"Lote de demonstração: {len(lote_demo)} endereços em "
      f"{len(set(RUA_DE[lote_demo].tolist()))} corredores")
print(f"Distância S-shape: {custo(seq_demo):.1f} m")
```

## 5. Roteirização por Colônia de Formigas (ACO)

Cada formiga constrói uma rota escolhendo o próximo endereço com probabilidade

$$p_{ij} = \frac{\tau_{ij}^{\alpha}\,\eta_{ij}^{\beta}}
                {\sum_{l \in \text{não visitados}} \tau_{il}^{\alpha}\,\eta_{il}^{\beta}}$$

em que $\tau$ é o feromônio depositado na aresta, $\eta = 1/d$ é a visibilidade
local, e $\alpha$ e $\beta$ ponderam memória coletiva e ganância. A cada
iteração o feromônio evapora à taxa $\rho$ e é reforçado nas arestas da melhor
rota encontrada (estratégia elitista).

```python
def aco_rota(locais, iteracoes=ACO_ITER, formigas=ACO_FORMIGAS, gerador=None,
             historico=False):
    '''Roteiriza um lote com colônia de formigas. Retorna (sequência, custo).'''
    g = gerador if gerador is not None else np.random.default_rng(SEMENTE)
    nos = np.concatenate(([DOCA], np.asarray(locais, dtype=int)))
    n = len(nos)
    if n <= 3:                      # nada a otimizar
        seq = [int(x) for x in nos] + [DOCA]
        c = custo(seq)
        return (seq, c, [c]) if historico else (seq, c)

    d = D[np.ix_(nos, nos)].copy()
    np.fill_diagonal(d, 1e-9)
    eta = 1.0 / d                   # visibilidade
    np.fill_diagonal(eta, 0.0)

    # Feromônio inicial calibrado pelo custo de uma solução gulosa
    ref = custo(rota_s_shape(locais))
    tau = np.full((n, n), 1.0 / max(ref, 1.0))

    melhor_seq, melhor_custo = None, np.inf
    curva = []

    for _ in range(iteracoes):
        for _ in range(formigas):
            visitado = np.zeros(n, dtype=bool)
            visitado[0] = True
            atual, caminho = 0, [0]
            for _ in range(n - 1):
                livres = np.flatnonzero(~visitado)
                peso = (tau[atual, livres] ** ACO_ALFA) * (eta[atual, livres] ** ACO_BETA)
                soma = peso.sum()
                prox = (g.choice(livres) if soma <= 0
                        else livres[g.choice(len(livres), p=peso / soma)])
                caminho.append(int(prox))
                visitado[prox] = True
                atual = int(prox)
            caminho.append(0)
            c = float(d[caminho[:-1], caminho[1:]].sum())
            if c < melhor_custo:
                melhor_custo, melhor_seq = c, list(caminho)

        tau *= (1.0 - ACO_RHO)                       # evaporação
        dep = ACO_Q / melhor_custo                   # reforço elitista
        for a, b in zip(melhor_seq[:-1], melhor_seq[1:]):
            tau[a, b] += dep
            tau[b, a] += dep
        curva.append(melhor_custo)

    seq = [int(nos[i]) for i in melhor_seq]
    return (seq, melhor_custo, curva) if historico else (seq, melhor_custo)


_, c_aco_demo, curva_demo = aco_rota(lote_demo, gerador=np.random.default_rng(1),
                                     historico=True)
print(f"Distância S-shape: {custo(seq_demo):8.1f} m")
print(f"Distância ACO:     {c_aco_demo:8.1f} m")
print(f"Redução:           {100*(1 - c_aco_demo/custo(seq_demo)):8.1f} %")
```

## 6. Formação de lotes por Algoritmo Genético

O cromossomo é um vetor de inteiros de comprimento igual ao número de pedidos da
onda: a posição $i$ guarda o lote atribuído ao pedido $i$. A aptidão soma a
distância estimada de todos os lotes e penaliza o excesso de capacidade.

**Decisão de projeto:** a aptidão usa o estimador rápido (*S-shape*), não o ACO.
Rodar a colônia de formigas para cada um dos milhares de cromossomos avaliados
seria proibitivo; o ACO entra depois, uma única vez, sobre os lotes já formados.
Essa separação entre *avaliar barato* e *refinar caro* é o que torna o pipeline
viável dentro da janela de segundos exigida pela operação.

```python
def lotes_fifo(pedidos):
    '''Linha de base: agrupa por ordem de chegada até encher o carrinho.'''
    lotes, atual, n = [], [], 0
    for i, p in enumerate(pedidos):
        if n + len(p) > CAP_LINHAS and atual:
            lotes.append(atual)
            atual, n = [], 0
        atual.append(i)
        n += len(p)
    if atual:
        lotes.append(atual)
    return lotes


def enderecos_do_lote(pedidos, indices):
    if not len(indices):
        return np.array([], dtype=int)
    return np.unique(np.concatenate([pedidos[i] for i in indices]))


def aptidao(crom, pedidos, n_lotes, tam):
    '''Distância total estimada + penalidade por estourar a capacidade.'''
    total, excesso = 0.0, 0
    for b in range(n_lotes):
        idx = np.flatnonzero(crom == b)
        if idx.size == 0:
            continue
        total += custo(rota_s_shape(enderecos_do_lote(pedidos, idx)))
        excesso += max(0, int(tam[idx].sum()) - CAP_LINHAS)
    return total + 1000.0 * excesso


def ag_batching(pedidos, n_lotes, geracoes=AG_GERACOES, pop=AG_POP, gerador=None):
    '''Evolui a atribuição pedido -> lote. Retorna (lotes, curva de aptidão).'''
    g = gerador if gerador is not None else np.random.default_rng(SEMENTE)
    n = len(pedidos)
    tam = np.array([len(p) for p in pedidos])

    # População inicial: um indivíduo semeado com o FIFO, o resto aleatório.
    populacao = g.integers(0, n_lotes, size=(pop, n))
    semente = np.zeros(n, dtype=int)
    for b, lote in enumerate(lotes_fifo(pedidos)):
        for i in lote:
            semente[i] = min(b, n_lotes - 1)
    populacao[0] = semente

    valores = np.array([aptidao(c, pedidos, n_lotes, tam) for c in populacao])
    curva = [valores.min()]

    for _ in range(geracoes):
        ordem = np.argsort(valores)
        nova = [populacao[i].copy() for i in ordem[:AG_ELITE]]      # elitismo
        while len(nova) < pop:
            # Seleção por torneio
            pais = []
            for _ in range(2):
                cand = g.choice(pop, size=AG_TORNEIO, replace=False)
                pais.append(populacao[cand[np.argmin(valores[cand])]])
            # Crossover uniforme
            mascara = g.random(n) < 0.5
            filho = np.where(mascara, pais[0], pais[1])
            # Mutação: realoca pedidos aleatórios para outro lote
            if g.random() < AG_P_MUT:
                quais = g.random(n) < (1.5 / n)
                filho[quais] = g.integers(0, n_lotes, size=quais.sum())
            nova.append(filho)

        populacao = np.array(nova)
        valores = np.array([aptidao(c, pedidos, n_lotes, tam) for c in populacao])
        curva.append(valores.min())

    melhor = populacao[np.argmin(valores)]
    lotes = [np.flatnonzero(melhor == b).tolist() for b in range(n_lotes)]
    return [l for l in lotes if l], curva
```

Um único lote, para tornar visível o que o ACO faz de diferente. A figura mostra
os dois percursos sobre a planta do armazém.

```python
g1 = np.random.default_rng(7)
pedidos_demo = gerar_pedidos(6, g1)
lote1 = np.unique(np.concatenate(pedidos_demo))

seq_ss = rota_s_shape(lote1)
c_ss = custo(seq_ss)
seq_ac, c_ac, curva_aco = aco_rota(lote1, gerador=np.random.default_rng(3),
                                   historico=True)

print(f"Endereços no lote: {len(lote1)} em "
      f"{len(set(RUA_DE[lote1].tolist()))} corredores")
print(f"S-shape : {c_ss:7.1f} m")
print(f"ACO     : {c_ac:7.1f} m   ({100*(1-c_ac/c_ss):.1f}% menor)")
```

```python
def desenhar(ax, seq, cor, titulo):
    # Endereços do armazém, ao fundo
    ax.scatter(XY[LOCAIS, 0], XY[LOCAIS, 1], s=1.2, c="#dddddd", zorder=1)
    # Percurso realmente caminhado
    for u, v in zip(seq[:-1], seq[1:]):
        pts = np.array(caminho_xy(u, v))
        ax.plot(pts[:, 0], pts[:, 1], color=cor, lw=1.7, alpha=0.85, zorder=2)
    # Paradas do lote e doca
    ax.scatter(XY[lote1, 0], XY[lote1, 1], s=42, c="#c62828",
               edgecolor="white", lw=0.8, zorder=4)
    ax.scatter(*XY[DOCA], s=130, marker="s", c="#2e7d32",
               edgecolor="white", lw=1.0, zorder=5)
    ax.set_title(titulo, fontsize=10.5)
    ax.set_xlabel("largura do CD (m)", fontsize=8)
    ax.set_aspect("equal")
    ax.tick_params(labelsize=7)

fig, axes = plt.subplots(1, 2, figsize=(11.5, 5.4), sharey=True)
desenhar(axes[0], seq_ss, "#1565c0", f"Linha de base — S-shape: {c_ss:.0f} m")
desenhar(axes[1], seq_ac, "#e07b00", f"Proposto — ACO: {c_ac:.0f} m "
                                     f"({100*(1-c_ac/c_ss):.1f}% menor)")
axes[0].set_ylabel("profundidade do CD (m)", fontsize=8)
fig.legend(handles=[
    Line2D([], [], marker="s", ls="", ms=8, mfc="#2e7d32", mec="white",
           label="Doca de expedição"),
    Line2D([], [], marker="o", ls="", ms=7, mfc="#c62828", mec="white",
           label="Endereços do lote"),
    Line2D([], [], color="#1565c0", lw=2, label="Rota S-shape"),
    Line2D([], [], color="#e07b00", lw=2, label="Rota ACO"),
], loc="lower center", ncol=4, frameon=False, fontsize=8.5)
fig.tight_layout(rect=[0, 0.06, 1, 1])
fig.savefig(BASE / "fig1_rotas.png", dpi=200, bbox_inches="tight", facecolor="white")
plt.show()
```

## 8. Teste 2 — experimento fatorial sobre uma onda completa

Uma onda de 60 pedidos é resolvida nas **quatro combinações** possíveis de
agrupamento (FIFO ou AG) e roteirização (S-shape ou ACO). Comparar as quatro,
e não apenas a linha de base contra a proposta, permite separar **quanto do
ganho vem de agrupar melhor** e **quanto vem de roteirizar melhor** — informação
gerencial relevante, porque as duas etapas têm custos de implantação diferentes.

O número de lotes é fixado no mesmo valor produzido pelo FIFO, de modo que a
comparação isole a *composição* dos lotes e não o uso de mais recursos.

```python
N_ONDA = 60
g2 = np.random.default_rng(2024)
pedidos = gerar_pedidos(N_ONDA, g2)
n_lotes = len(lotes_fifo(pedidos))

print(f"Onda: {N_ONDA} pedidos, {sum(len(p) for p in pedidos)} linhas, "
      f"{n_lotes} lotes de até {CAP_LINHAS} linhas")

t0 = time.perf_counter()
lotes_ag, curva_ag = ag_batching(pedidos, n_lotes,
                                 gerador=np.random.default_rng(11))
tempo_ag = time.perf_counter() - t0
print(f"AG: {AG_GERACOES} gerações x {AG_POP} indivíduos em {tempo_ag:.1f} s")
```

```python
def avaliar(lotes, roteirizador, gerador):
    '''Distância total da onda para uma dada combinação agrupamento+rota.'''
    total = 0.0
    for lote in lotes:
        end = enderecos_do_lote(pedidos, lote)
        if roteirizador == "s-shape":
            total += custo(rota_s_shape(end))
        else:
            total += aco_rota(end, gerador=gerador)[1]
    return total

configs = {
    "FIFO + S-shape\n(linha de base)": (lotes_fifo(pedidos), "s-shape"),
    "FIFO + ACO":                      (lotes_fifo(pedidos), "aco"),
    "AG + S-shape":                    (lotes_ag, "s-shape"),
    "AG + ACO\n(proposto)":            (lotes_ag, "aco"),
}

t0 = time.perf_counter()
distancias = {nome: avaliar(lotes, rot, np.random.default_rng(5))
              for nome, (lotes, rot) in configs.items()}
tempo_total = time.perf_counter() - t0

base = distancias["FIFO + S-shape\n(linha de base)"]
tabela = pd.DataFrame({
    "configuração": [k.replace("\n", " ") for k in distancias],
    "distância total (m)": [round(v, 1) for v in distancias.values()],
    "distância/pedido (m)": [round(v / N_ONDA, 1) for v in distancias.values()],
    "redução vs. base (%)": [round(100 * (1 - v / base), 1) for v in distancias.values()],
})
print(f"Tempo de avaliação das 4 configurações: {tempo_total:.1f} s\n")
tabela
```

```python
fig, axes = plt.subplots(1, 2, figsize=(11.5, 4.0))

# (a) convergência
ax = axes[0]
ax.plot(curva_ag, color="#e07b00", lw=1.8, label="AG — melhor aptidão (onda inteira)")
ax.set_xlabel("geração", fontsize=9)
ax.set_ylabel("distância estimada da onda (m)", fontsize=9, color="#e07b00")
ax.tick_params(axis="y", labelcolor="#e07b00", labelsize=8)
ax.tick_params(axis="x", labelsize=8)
ax2 = ax.twinx()
ax2.plot(curva_aco, color="#1565c0", lw=1.8, label="ACO — melhor rota (lote do Teste 1)")
ax2.set_ylabel("distância da rota (m)", fontsize=9, color="#1565c0")
ax2.tick_params(axis="y", labelcolor="#1565c0", labelsize=8)
ax.set_title("(a) Convergência dos dois algoritmos", fontsize=10.5)
linhas = ax.get_lines() + ax2.get_lines()
ax.legend(linhas, [l.get_label() for l in linhas], fontsize=8, frameon=False,
          loc="upper right")
ax.grid(alpha=0.25)

# (b) comparativo das 4 configurações
ax = axes[1]
nomes = list(distancias)
vals = [distancias[n] for n in nomes]
cores = ["#9e9e9e", "#90caf9", "#ffcc80", "#e07b00"]
barras = ax.bar(range(len(nomes)), vals, color=cores, edgecolor="#555555", lw=0.7)
for i, (b, v) in enumerate(zip(barras, vals)):
    ax.text(b.get_x() + b.get_width() / 2, v + max(vals) * 0.012,
            f"{v:,.0f} m".replace(",", "."), ha="center", fontsize=8.5)
    if i:
        ax.text(b.get_x() + b.get_width() / 2, v * 0.5,
                f"−{100*(1-v/base):.1f}%", ha="center", fontsize=9,
                fontweight="bold", color="#333333")
ax.set_xticks(range(len(nomes)))
ax.set_xticklabels(nomes, fontsize=8)
ax.set_ylabel("distância total da onda (m)", fontsize=9)
ax.set_title("(b) Contribuição isolada de cada componente", fontsize=10.5)
ax.set_ylim(0, max(vals) * 1.12)
ax.tick_params(axis="y", labelsize=8)
ax.grid(axis="y", alpha=0.25)

fig.tight_layout()
fig.savefig(BASE / "fig2_convergencia.png", dpi=200, bbox_inches="tight",
            facecolor="white")
plt.show()
```

## 9. Teste 3 — escalabilidade e estabilidade do ganho

Um ganho medido em uma única onda pode ser sorte. Aqui o experimento é repetido
em **três tamanhos de onda**, com **réplicas independentes** (sementes distintas),
para verificar se a vantagem se sustenta e qual é sua dispersão.

```python
TAMANHOS = [30, 60, 120]
N_REPLICAS = 8

registros = []
t0 = time.perf_counter()
for n_ped in TAMANHOS:
    for rep in range(N_REPLICAS):
        gr = np.random.default_rng(1000 + 97 * n_ped + rep)
        peds = gerar_pedidos(n_ped, gr)
        nl = len(lotes_fifo(peds))

        # Linha de base
        d_base = sum(custo(rota_s_shape(enderecos_do_lote(peds, l)))
                     for l in lotes_fifo(peds))

        # Proposto (parâmetros reduzidos para manter o tempo do teste razoável)
        lts, _ = ag_batching(peds, nl, geracoes=30, pop=24, gerador=gr)
        d_prop = sum(aco_rota(enderecos_do_lote(peds, l), iteracoes=30,
                              formigas=14, gerador=gr)[1] for l in lts)

        registros.append({"pedidos": n_ped, "réplica": rep,
                          "base (m)": d_base, "proposto (m)": d_prop,
                          "redução (%)": 100 * (1 - d_prop / d_base)})
tempo_t3 = time.perf_counter() - t0

df = pd.DataFrame(registros)
resumo = df.groupby("pedidos").agg(
    base_media=("base (m)", "mean"),
    proposto_medio=("proposto (m)", "mean"),
    reducao_media=("redução (%)", "mean"),
    reducao_dp=("redução (%)", "std"),
).round(2)
print(f"Teste 3 concluído em {tempo_t3:.0f} s "
      f"({len(TAMANHOS)*N_REPLICAS} execuções)\n")
resumo
```

```python
fig, axes = plt.subplots(1, 2, figsize=(11.0, 4.0))

# (a) distância média: base x proposto
ax = axes[0]
larg, pos = 0.36, np.arange(len(TAMANHOS))
ax.bar(pos - larg/2, resumo["base_media"], larg, label="Linha de base (FIFO + S-shape)",
       color="#9e9e9e", edgecolor="#555555", lw=0.7)
ax.bar(pos + larg/2, resumo["proposto_medio"], larg, label="Proposto (AG + ACO)",
       color="#e07b00", edgecolor="#555555", lw=0.7)
ax.set_xticks(pos)
ax.set_xticklabels([f"{t} pedidos" for t in TAMANHOS], fontsize=8.5)
ax.set_ylabel("distância média da onda (m)", fontsize=9)
ax.set_title("(a) Distância percorrida por tamanho de onda", fontsize=10.5)
ax.legend(fontsize=8, frameon=False)
ax.grid(axis="y", alpha=0.25)
ax.tick_params(axis="y", labelsize=8)

# (b) redução percentual, com dispersão entre réplicas
ax = axes[1]
ax.bar(pos, resumo["reducao_media"], 0.5, yerr=resumo["reducao_dp"], capsize=6,
       color="#1565c0", edgecolor="#555555", lw=0.7,
       error_kw=dict(ecolor="#333333", lw=1.2))
for i, (m, s) in enumerate(zip(resumo["reducao_media"], resumo["reducao_dp"])):
    ax.text(i, m + s + 0.5, f"{m:.1f}%", ha="center", fontsize=9, fontweight="bold")
ax.set_xticks(pos)
ax.set_xticklabels([f"{t} pedidos" for t in TAMANHOS], fontsize=8.5)
ax.set_ylabel("redução da distância (%)", fontsize=9)
ax.set_title(f"(b) Ganho médio ± desvio-padrão ({N_REPLICAS} réplicas)", fontsize=10.5)
ax.set_ylim(0, max(resumo["reducao_media"] + resumo["reducao_dp"]) * 1.30)
ax.grid(axis="y", alpha=0.25)
ax.tick_params(axis="y", labelsize=8)

fig.tight_layout()
fig.savefig(BASE / "fig3_escalabilidade.png", dpi=200, bbox_inches="tight",
            facecolor="white")
plt.show()
```
